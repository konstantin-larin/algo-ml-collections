```python
import numpy as np
```

```python
data = np.loadtxt('data.csv', delimiter=',')
x = data[:, 0]
y = data[:, 1]
```

```python
sinx = np.sin(x)
lnx = np.log(x)

X = np.vstack([sinx ** 2, 2 * sinx * lnx, lnx ** 2, x ** 2]).T
```

```python
params, *_ = np.linalg.lstsq(X, y, rcond=None)
print(params)
```

```python
p1, p2, p3, p4 = params
```

```python
A = np.sqrt(max(p1, 0.0))
if A < 1e-12:
    B = np.sqrt(max(p3, 0.0))
else:
    B = p2 / A

B_alt = np.sqrt(max(p3, 0.0))

if B_alt > 0:
    B = 0.5 * (B + B_alt)

C = p4
print(A, B, C)
```

Решим с помощью МНК аналитически

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures

train_data = np.array([list(map(float, input().split(sep='\t'))) for _ in range(1000)])
X_train = train_data[:, :5]
y_train = train_data[:, 5] 
X_test = np.array([list(map(float, input().split(sep='\t'))) for _ in range(1000)])


poly = PolynomialFeatures(degree=2, include_bias=True)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
w, _, _, _ = np.linalg.lstsq(X_train_poly, y_train, rcond=None)
w = w.reshape(-1, 1)

y_pred = X_test_poly.dot(w)

for num in y_pred:
    print(num[0])
```

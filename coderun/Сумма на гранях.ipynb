```python
import sys
import numpy as np


def main():
    """
    Пример ввода и вывода числа n, где -10^9 < n < 10^9:
    n = int(input())
    print(n)
    """
    unique_numbers, counts = np.unique(np.fromstring(input(), dtype=int, sep=' '), return_counts=True)
    probas = counts / 6    
    reverse_probas = 1 - probas
    k = int(input())

    math_expectations = [np.sum(unique_numbers * probas), ]    
    for _ in range(1, k):         
        math_expectations.append(
            np.sum(unique_numbers * probas * reverse_probas)
        )

    return np.sum(math_expectations)    


if __name__ == '__main__':
    print(main())
```
